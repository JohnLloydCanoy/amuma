```python
# Define the mapping for Current Year Level to Age
year_to_age = {
    "First Year": 19,
    "Second Year": 20,
    "Third Year": 21,
    "Fourth Year": 22,
    "Forth Year": 22,  # Added to cover potential typos
    "Fifth Year": 23,
}

# Map the Current Year Level to ages and fill ONLY the null values in 'Age'
mapped_ages = df["Current Year Level"].map(year_to_age)
df["Age"] = df["Age"].fillna(mapped_ages)

# Check if there are any nulls left in Age
print("Remaining missing Age values:", df["Age"].isnull().sum())
```

```python
import numpy as np

# 1. Safely grab the exact column names using their index (2 and 3)
col_dept = df.columns[2]
col_prog = df.columns[3]

# 2. List of text variations that basically mean "blank"
na_keywords = ['n/a', 'na', 'none', '-', '', 'not applicable', 'nan']

print("Standardizing College and Program columns...")

# 3. Loop through both columns to clean them
for col in [col_dept, col_prog]:
    # Find rows where the typed text matches our keywords (ignores uppercase/lowercase and spaces)
    text_matches = df[col].astype(str).str.lower().str.strip().isin(na_keywords)

    # Convert those weird text variations into actual Pandas Nulls (NaN)
    df.loc[text_matches, col] = np.nan

    # Now fill ALL nulls (both the original blanks and the ones we just converted) with the standard "N/A"
    df[col] = df[col].fillna("N/A")

# 4. Verify it worked (Both should now say 0 missing values!)
print(f"Missing values in College: {df[col_dept].isnull().sum()}")
print(f"Missing values in Program: {df[col_prog].isnull().sum()}")
```

```python
# A list of words that usually mean "Not Applicable" or are manually typed blanks
na_keywords = ['n/a', 'na', 'none', 'nothing', '-', 'not applicable', 'no answer', '']

print("--- Scanning for 'N/A' or similar variations ---")
found_any = False

# We use enumerate to get both the column index (i) and the name (col)
for i, col in enumerate(df.columns):

    # 1. Grab the column by its exact numerical position (i) to avoid duplicate name errors
    # 2. Drop actual pandas nulls for the check
    # 3. Convert to string, make lowercase, and remove extra spaces
    text_data = df.iloc[:, i].dropna().astype(str).str.lower().str.strip()

    # Check if the text exactly matches any of our NA keywords
    matches = text_data[text_data.isin(na_keywords)]

    # If we found any matches, print the summary for this column
    if not matches.empty:
        found_any = True
        count = len(matches)
        unique_found = matches.unique()  # Shows exactly which variations were used

        # Truncate long column names for a cleaner print output
        display_col = col if len(col) < 60 else col[:57] + "..."

        # Print the column index along with the name to help you locate duplicates
        print(f"Column {i}: {display_col}")
        print(f"  - Found {count} instances.")
        print(f"  - Variations found: {list(unique_found)}\n")

if not found_any:
    print("No 'N/A' or similar variations were found in the dataset.")
```
