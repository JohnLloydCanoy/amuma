```python
# Define the mapping for Current Year Level to Age
year_to_age = {
    "First Year": 19,
    "Second Year": 20,
    "Third Year": 21,
    "Fourth Year": 22,
    "Forth Year": 22,  # Added to cover potential typos
    "Fifth Year": 23,
}

# Map the Current Year Level to ages and fill ONLY the null values in 'Age'
mapped_ages = df["Current Year Level"].map(year_to_age)
df["Age"] = df["Age"].fillna(mapped_ages)

# Check if there are any nulls left in Age
print("Remaining missing Age values:", df["Age"].isnull().sum())
```

```python
import numpy as np

# 1. Safely grab the exact column names using their index (2 and 3)
col_dept = df.columns[2]
col_prog = df.columns[3]

# 2. List of text variations that basically mean "blank"
na_keywords = ['n/a', 'na', 'none', '-', '', 'not applicable', 'nan']

print("Standardizing College and Program columns...")

# 3. Loop through both columns to clean them
for col in [col_dept, col_prog]:
    # Find rows where the typed text matches our keywords (ignores uppercase/lowercase and spaces)
    text_matches = df[col].astype(str).str.lower().str.strip().isin(na_keywords)

    # Convert those weird text variations into actual Pandas Nulls (NaN)
    df.loc[text_matches, col] = np.nan

    # Now fill ALL nulls (both the original blanks and the ones we just converted) with the standard "N/A"
    df[col] = df[col].fillna("N/A")

# 4. Verify it worked (Both should now say 0 missing values!)
print(f"Missing values in College: {df[col_dept].isnull().sum()}")
print(f"Missing values in Program: {df[col_prog].isnull().sum()}")
```

```python
# A list of words that usually mean "Not Applicable" or are manually typed blanks
na_keywords = ['n/a', 'na', 'none', 'nothing', '-', 'not applicable', 'no answer', '']

print("--- Scanning for 'N/A' or similar variations ---")
found_any = False

# We use enumerate to get both the column index (i) and the name (col)
for i, col in enumerate(df.columns):

    # 1. Grab the column by its exact numerical position (i) to avoid duplicate name errors
    # 2. Drop actual pandas nulls for the check
    # 3. Convert to string, make lowercase, and remove extra spaces
    text_data = df.iloc[:, i].dropna().astype(str).str.lower().str.strip()

    # Check if the text exactly matches any of our NA keywords
    matches = text_data[text_data.isin(na_keywords)]

    # If we found any matches, print the summary for this column
    if not matches.empty:
        found_any = True
        count = len(matches)
        unique_found = matches.unique()  # Shows exactly which variations were used

        # Truncate long column names for a cleaner print output
        display_col = col if len(col) < 60 else col[:57] + "..."

        # Print the column index along with the name to help you locate duplicates
        print(f"Column {i}: {display_col}")
        print(f"  - Found {count} instances.")
        print(f"  - Variations found: {list(unique_found)}\n")

if not found_any:
    print("No 'N/A' or similar variations were found in the dataset.")
```

```python
import os
import matplotlib.patheffects as pe
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Turn on Seaborn styling
sns.set_theme(style="white")

# 2. Load the dataset
cleaned_df = pd.read_csv("rawdata/Cleaned_Data.csv")

# 3. Grab Column 19 (Index 18)
feature_col = cleaned_df.columns[18]

# 4. SPLIT MULTIPLE ANSWERS
# (Assuming comma-separated; switch to ";" if normalized with semicolons during cleaning)
separated_features = (
    cleaned_df[feature_col]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
)

# Clean up accidental spaces and remove empty entries
separated_features = separated_features.str.strip()
separated_features = separated_features[separated_features != ""]

# 5. Count the answers
feature_counts = separated_features.value_counts().reset_index()
feature_counts.columns = ["Feature", "Count"]

# Calculate the TOTAL number of votes for percentages
total_votes = feature_counts["Count"].sum()

# 6. Set up the figure (Taller to comfortably fit all options)
plt.figure(figsize=(10, 8), dpi=300)

# The Extended Calico Theme Palette
extended_calico_colors = [
    "#E87A3E", "#1D1D1B", "#F5EBE1", "#5C3A21", "#E2A36B",
    "#A65D37", "#D9C3A9", "#8C4F2B",
]

colors = [
    extended_calico_colors[i % len(extended_calico_colors)]
    for i in range(len(feature_counts))
]

# 7. Create the Horizontal Bar Chart
ax = sns.barplot(
    data=feature_counts,
    x="Count",
    y="Feature",
    palette=colors,
    edgecolor="black",
    linewidth=1.5,
)

# Give 15% breathing room on the right for the numbers
max_count = feature_counts["Count"].max()
plt.xlim(0, max_count * 1.15)

# 8. Add the Exact Numbers and Percentages
for patch in ax.patches:
    width = patch.get_width()
    if width > 0:
        # A) Put the COUNT to the right of the bar
        plt.text(
            width + (max_count * 0.015),
            patch.get_y() + patch.get_height() / 2,
            str(int(width)),
            ha="left",
            va="center",
            fontsize=11,
            fontweight="bold",
        )

        # B) Put the PERCENTAGE INSIDE the bar (if wide enough)
        if width > (max_count * 0.08):
            percentage = (width / total_votes) * 100
            plt.text(
                width / 2,
                patch.get_y() + patch.get_height() / 2,
                f"{percentage:.1f}%",
                ha="center",
                va="center",
                fontsize=11,
                fontweight="bold",
                color="black",
                path_effects=[pe.withStroke(linewidth=3, foreground="white")],
            )

# 9. Add title and styling
plt.title(
    "Preferred Amuma AI Features\n(Respondents could choose multiple)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Number of Votes", fontsize=12)
plt.ylabel("", fontsize=12)

plt.yticks(fontsize=11)
plt.grid(axis="x", linestyle="--", alpha=0.5)

# 10. Save to the specific folder
save_folder = "data_visualization"
os.makedirs(save_folder, exist_ok=True)

file_path = os.path.join(save_folder, "Amuma_AI_Features_Bar.png")
plt.savefig(file_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Success! The Features chart has been saved to: {file_path}")
```
