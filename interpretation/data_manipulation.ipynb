```python
import pandas as pd 
df = pd.read_csv("rawdata\RawData.csv")
df.columns = [col[:30] + '...' if len(col) > 30 else col for col in df.columns] 
df.info()
```

```python
# The column numbers we want to fill
columns_to_fill = [28, 29]

for col_idx in columns_to_fill:
    col_name = df.columns[col_idx]

    # Fill the empty (null) values with the string "N/A"
    df[col_name] = df[col_name].fillna("N/A")

    # Verify the new non-null count (should print 709 for both)
    non_null_count = df[col_name].notnull().sum()
    print(f"Column {col_idx} now has {non_null_count} non-null values.")
```
